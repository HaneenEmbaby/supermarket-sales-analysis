```python
df1 = df[(df['Price'] > 5000) | (df['Rating'] > 4.5)]
df1
```

```python
df = df.sort_values('Price', ascending=False)
df
```

```python
df.isna().sum()
```

```python
df.fillna(df['Price'].mean(), inplace=True)
df.fillna(df['Stock'].mean(), inplace=True)
```

```python
df['In_stock'] = df['Stock'].apply(lambda x: 'True' if x > 0 else 'False')
df['Total_value'] = df['Price'] * df['Stock']
```

```python
df2 = df[(df["Price"] > 3000) & (df["Rating"] >= 4.0)]
df2
```

```python
df['Stock'].mean()
df['Stock'].min()
df['Stock'].max()
df['Stock'].sum()
```

```python
df['Price_Category']= df['Price'].apply(lambda x: 'Expensive' if x > 10000 else ('Medium' if x >= 3000 else 'Cheap'))
df
```

```python
df.rename(columns={'Product_Name': 'Name'}, inplace=True)
df
```

```python
df.drop(['Product_ID'], axis=1)
```
